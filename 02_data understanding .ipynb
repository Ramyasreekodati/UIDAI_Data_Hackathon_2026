```python
for df in [enrol, demo, bio]:
    df['state'] = df['state'].str.strip().str.title()
    df['district'] = df['district'].str.strip().str.title()
    df['date'] = pd.to_datetime(df['date'], dayfirst=True)
```

converting state,district,date into valid values no extra space and same formate like title

```python
merged_1 = enrol.merge(
    demo,
    on=['date', 'state', 'district', 'pincode'],
    how='outer'
)
```

we starting merge of enrollment and demographic data = merged_1 

```python
aadhar_df = merged_1.merge(
    bio,
    on=['date', 'state', 'district', 'pincode'],
    how='outer'
)
```

now merged_1 is added of biometric so all 3 datasets are merged

and we found some duplicate values to 

```python
keys = ['date', 'state', 'district', 'pincode']

enrol_dups = enrol.duplicated(subset=keys, keep=False).sum()
demo_dups  = demo.duplicated(subset=keys, keep=False).sum()
bio_dups   = bio.duplicated(subset=keys, keep=False).sum()

enrol_dups, demo_dups, bio_dups
```

we are spliting the data into 3 parts like enrol_duplicates , demographic duplicated and biometric duplicated

```python
enrol_clean = (
    enrol
    .groupby(keys, as_index=False)
    .agg({
        'age_0_5': 'sum',
        'age_5_17': 'sum',
        'age_18_greater': 'sum'
    })
)
```

This code groups the enrollment data by key fields like date and location, and then sums the enrollment counts for different age groups (0–5, 5–17, and 18+).
The result is a cleaned dataset with total enrollments per age group for each group.

```python
demo_clean = (
    demo
    .groupby(keys, as_index=False)
    .agg({
        'demo_age_5_17': 'sum',
        'demo_age_17_': 'sum'
    })
)
```

This code groups the demographic data by key fields like date and location, and then sums the demographic counts for different age groups (0–5, 5–17, and 18+).
The result is a cleaned dataset with total enrollments per age group for each group.

```python
bio_clean = (
    bio
    .groupby(keys, as_index=False)
    .agg({
        'bio_age_5_17': 'sum',
        'bio_age_17_': 'sum'
    })
)
```

This code groups the biometric data by key fields like date and location, and then sums the biometric counts for different age groups (0–5, 5–17, and 18+).
The result is a cleaned dataset with total enrollments per age group for each group.

```python
enrol_clean.duplicated(subset=keys).sum()  
```

```python
aadhar_df = (
    enrol_clean
    .merge(demo_clean, on=keys, how='outer')
    .merge(bio_clean, on=keys, how='outer')
)
```

```python
aadhar_df.duplicated(subset=keys).sum()
```

there is no duplicates now

```python
aadhar_df.isna().sum()
```

```python
activity_cols = [
    'age_0_5','age_5_17','age_18_greater',
    'demo_age_5_17','demo_age_17_',
    'bio_age_5_17','bio_age_17_'
]

aadhar_df[activity_cols] = aadhar_df[activity_cols].fillna(0).astype(int)
```

we have observed some null values in that place we fill 0

```python
aadhar_df.duplicated(subset=keys).sum()
```

```python
aadhar_df['total_activity'] = aadhar_df[activity_cols].sum(axis=1)
aadhar_df['total_activity'].describe()
```

main we are adding all the enrolment ,demographic and biometric totals into one = total_activity
