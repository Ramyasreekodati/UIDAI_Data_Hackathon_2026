```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats 
```

```python
Aadhaar_df=pd.read_csv(r'C:\Users\Lenovo\Desktop\UIDAI_Data_Hackathon_2026\data\combined\cleaned_aadhaar_data.csv')
```

```python
Aadhaar_df['date'] = pd.to_datetime(Aadhaar_df['date'])

Aadhaar_df['day'] = Aadhaar_df['date'].dt.day
Aadhaar_df['weekday'] = Aadhaar_df['date'].dt.weekday      
Aadhaar_df['weekday_name'] = Aadhaar_df['date'].dt.day_name()
Aadhaar_df['is_weekend'] = Aadhaar_df['weekday'].isin([5,6]).astype(int)

Aadhaar_df['month_num'] = Aadhaar_df['date'].dt.month
Aadhaar_df['quarter'] = Aadhaar_df['date'].dt.quarter
```

```python
Aadhaar_df.head()
```

```python
Aadhaar_df['enrollment_total'] = Aadhaar_df[
    ['age_0_5','age_5_17','age_18_greater']
].sum(axis=1)

Aadhaar_df['demographic_total'] = Aadhaar_df[
    ['demo_age_5_17','demo_age_17_']
].sum(axis=1)

Aadhaar_df['biometric_total'] = Aadhaar_df[
    ['bio_age_5_17','bio_age_17_']
].sum(axis=1)
```

combining all ages into one total

```python
Aadhaar_df.head()
```

```python
Aadhaar_df['bio_share'] = Aadhaar_df['biometric_total'] / Aadhaar_df['total_activity']
Aadhaar_df['demo_share'] = Aadhaar_df['demographic_total'] / Aadhaar_df['total_activity']
Aadhaar_df['enroll_share'] = Aadhaar_df['enrollment_total'] / Aadhaar_df['total_activity']
```

what ever the total are there we making some percentile of each group

```python
Aadhaar_df.head()
```

```python
Aadhaar_df[['bio_share','demo_share','enroll_share']] = (
    Aadhaar_df[['bio_share','demo_share','enroll_share']].fillna(0)
)
```

if there is any any null value fill with 0

```python
Aadhaar_df['child_activity'] = Aadhaar_df['age_0_5'] + Aadhaar_df['age_5_17']
Aadhaar_df['adult_activity'] = Aadhaar_df['age_18_greater']
```

considering 0 -17 as child activity and 17+ as adult

```python
pincode_mean = Aadhaar_df.groupby('pincode')['total_activity'].mean()

Aadhaar_df['pincode_avg_activity'] = Aadhaar_df['pincode'].map(pincode_mean)
```

First, this code calculates the average Aadhaar activity for each pincode.
Then, it assigns that average value back to every record of the same pincode.
This helps compare today’s activity with what is normally expected for that area

```python
Aadhaar_df.head()
```

```python
Aadhaar_df = Aadhaar_df.sort_values(['state','district','pincode','date'])

Aadhaar_df['lag_1_day'] = Aadhaar_df.groupby(
    ['state','district','pincode']
)['total_activity'].shift(1)

Aadhaar_df['lag_7_day'] = Aadhaar_df.groupby(
    ['state','district','pincode']
)['total_activity'].shift(7)
```

First, the data is sorted by location and date so that records are in time order.
Then, for each pincode, the previous day’s activity and the activity from 7 days ago are added as new columns.
This helps compare today’s Aadhaar activity with recent past trends and detect changes or spikes.

```python
Aadhaar_df['rolling_7_mean'] = Aadhaar_df.groupby(
    ['state','district','pincode']
)['total_activity'].transform(lambda x: x.rolling(7).mean())
```

This code calculates the average Aadhaar activity over the last 7 days for each pincode.
It smooths daily ups and downs and shows the normal weekly activity level for that location.

```python
threshold = Aadhaar_df['total_activity'].quantile(0.99)

Aadhaar_df['high_activity_flag'] = (
    Aadhaar_df['total_activity'] > threshold
).astype(int)
```

It flags the top 1% highest activity days to detect abnormal spikes.


This code finds the activity level that only the top 1% of days exceed.
Any day with activity higher than this level is marked as a high-activity day.
It helps quickly identify unusual spikes or special event days.

```python
Aadhaar_df['state_code'] = Aadhaar_df['state'].astype('category').cat.codes
Aadhaar_df['district_code'] = Aadhaar_df['district'].astype('category').cat.codes
```
